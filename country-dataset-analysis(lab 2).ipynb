## 2. Data cleaning

```python
df["SNO"] = df["SNO"].astype("string").str.strip().str.upper()

# Remove rows without region names
df = df.dropna(subset=["SNO"])

# Remove the total summary row
df = df[df["SNO"] != "TOTAL"].copy()

# Convert numerical columns to numeric data types
numeric_columns = [
    "TOTALSCHOOLS",
    "TotalSchoolsInstalled",
    "SchoolsPending",
    "InstalledLDDs",
    "InstalledTDDs"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Convert installation percentage from text to a number
df["PercentageInstalled"] = pd.to_numeric(
    df["PercentageInstalled"]
      .astype(str)
      .str.replace("%", "", regex=False),
    errors="coerce"
)
# Remove rows containing missing values in the analysis columns
required_columns = numeric_columns + ["PercentageInstalled"]
df = df.dropna(subset=required_columns)

# Remove duplicate regions, if any
df = df.drop_duplicates(subset=["SNO"])



print("Cleaned dataset shape:", df.shape)
display(df.head())
print("\nMissing values after cleaning:")
print(df.isnull().sum())
print("\nData types after cleaning:")
print(df.dtypes)
```

## 3. Experiment 1: Anomaly detection

**3.1 objective**

Anomaly detection identifies observations that are significantly different from the general pattern of the dataset.

In this experiment, a region might be flagged because it has:
- An unusually high or low number of schools.
- An unusually high number of installed devices.
- A large number of schools still awaiting installation.
- A combination of school installation statistics that differs from those of other regions.
We will use the Isolation Forest algorithm, which identifies unusual observations by isolating them from the rest of the data.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest

# Select features for anomaly detection
features = [
    "TOTALSCHOOLS",
    "TotalSchoolsInstalled",
    "SchoolsPending",
    "PercentageInstalled",
    "InstalledLDDs",
    "InstalledTDDs"
]

X = df[features]

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create the Isolation Forest model
model = IsolationForest(
    contamination=0.10,
    random_state=42
)

# Fit the model and predict anomalies
df["Anomaly"] = model.fit_predict(X_scaled)

# Calculate anomaly scores
df["AnomalyScore"] = model.decision_function(X_scaled)

# Convert model predictions into readable labels
df["AnomalyStatus"] = df["Anomaly"].map({
    1: "Normal",
    -1: "Anomaly"
})

# Display the results
display(df[
    ["SNO", "AnomalyStatus", "AnomalyScore"]
].sort_values("AnomalyScore"))

# Display detected anomalies
anomalies = df[df["Anomaly"] == -1]

print("\nDetected anomalies:")
display(anomalies[
    ["SNO"] + features + ["AnomalyScore"]
].sort_values("AnomalyScore"))
```

```python
import matplotlib.pyplot as plt

colors = df["Anomaly"].map({
    1: "blue",
    -1: "red"
})

plt.figure(figsize=(12, 6))

plt.scatter(
    df["TOTALSCHOOLS"],
    df["PercentageInstalled"],
    c=colors,
    s=70
)

for _, row in df[df["Anomaly"] == -1].iterrows():
    plt.annotate(
        row["SNO"],
        (row["TOTALSCHOOLS"], row["PercentageInstalled"]),
        fontsize=8
    )

plt.xlabel("Total Number of Schools")
plt.ylabel("Percentage of Schools Installed")
plt.title("Anomaly Detection in the Country Schools Dataset")
plt.grid(True, alpha=0.3)
plt.show()
```

The Isolation Forest experiment identifies regions with unusual combinations of school counts, installation progress, and device statistics. Turkana is notable for its low installation progress, while Kitui and Makueni have high installation progress. Kakamega and Machakos have distinctive combinations of school and device counts.

An anomaly does not automatically mean that a record is bad. It indicates that the further examination is recomended. The contamination=0.10 setting specifies an expected anomaly proportion of approximately 10%; it is a model setting, not a measured rate of data errors.

## 4. Experiment 2: Cluster analysis

**4.1 Objective**

Cluster analysis groups regions according to similarities in their school installation statistics. Unlike anomaly detection, clustering does not primarily search for unusual observations. Instead, it discovers groups of regions with similar characteristics.

We will use K-Means clustering to divide the 47 regions into three clusters.

The expected groups are:
- Regions with relatively high numbers of schools and substantial installation progress.
- Regions with relatively few schools and high installation progress.
- Regions with relatively low installation progress and more schools awaiting installation.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Select features for clustering
features = [
    "TOTALSCHOOLS",
    "TotalSchoolsInstalled",
    "SchoolsPending",
    "PercentageInstalled",
    "InstalledLDDs",
    "InstalledTDDs"
]

X = df[features]

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create the K-Means model with three clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Assign a cluster to every region
df["Cluster"] = kmeans.fit_predict(X_scaled)

# Display the region and its cluster
display(df[["SNO", "Cluster"]].sort_values("Cluster"))

# Calculate average feature values for each cluster
cluster_summary = df.groupby("Cluster")[features].mean()

print("\nCluster summary:")
display(cluster_summary.round(2))

# Evaluate cluster separation
score = silhouette_score(X_scaled, df["Cluster"])

print("\nSilhouette score:", round(score, 3))
```

**Visualize the clusters**

```python
plt.figure(figsize=(12, 6))

for cluster in sorted(df["Cluster"].unique()):
    subset = df[df["Cluster"] == cluster]

    plt.scatter(
        subset["TOTALSCHOOLS"],
        subset["PercentageInstalled"],
        s=70,
        label="Cluster " + str(cluster)
    )

plt.xlabel("Total Number of Schools")
plt.ylabel("Percentage of Schools Installed")
plt.title("K-Means Clustering of Country Schools")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

**4.2 Clusters interpretation**



Cluster 0 (16 regions) — Larger school systems with good installation progress.
This cluster has the highest average number of schools (approximately 743) and an average installation rate of 72.69%. Regions include Kitui, Homa Bay, Migori, Kakamega, Bungoma, and Nakuru.

Interpretation: These regions have relatively large school systems and high overall device counts. Despite their installation progress, an average of approximately 200 schools per region remains pending. They may benefit from continued installation and monitoring.


Cluster 1 (17 regions)— Smaller school systems with relatively high installation progress. This cluster has an average of approximately 310 schools per region and the highest average installation percentage, 80.36%. Examples include Tana River, Taita Taveta, Vihiga, Nyamira, Mombasa, Lamu, and Embu.

Interpretation: These regions generally have fewer schools and higher installation percentages. Their average of approximately 65 pending schools is the lowest of the three clusters, suggesting that they are relatively further along in installation.


Cluster 2 (14 regions) — Lower installation progress. This cluster has an average of approximately 327 schools per region, but only 127.79 installed schools on average. Its average installation percentage is 33.92%, and approximately 199 schools per region remain pending.

Examples include Turkana, West Pokot, Garissa, Mandera, Wajir, Marsabit, Nyeri, and Samburu.

Interpretation: These regions have substantially lower average installation progress. They could be prioritized for further investigation, resource allocation, and installation support.
